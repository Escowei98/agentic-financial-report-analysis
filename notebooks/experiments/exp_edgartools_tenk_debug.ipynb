# Experiment: edgartools TenK Object Inspection

**Goal:** Understand what attributes/methods the TenK object provides for section extraction.

```python
from edgar import Company, set_identity
from src.common.config import load_config

config = load_config()
set_identity(config["sec_edgar_email"])

company = Company("AAPL")
filing = company.get_filings(form="10-K").latest()
tenk = filing.obj()

# Show type and all public attributes
print(f"Type: {type(tenk)}")
print(f"\nPublic attributes:")
for attr in sorted(dir(tenk)):
    if not attr.startswith("_"):
        print(f"  {attr}")
```
